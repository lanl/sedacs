```python
x, a, b, c, d, e, f, g, h = symbols('x a b c d e f g h')

def quad(x,a,b,c):
    return a + b * x + c * x * x

y1 = quad(x, a,b,c)
y2 = quad(y1,d,e,f)

# y1 = (a * x + b) ** 2
# y2 = (d * y1 + e * x + f) ** 2

p = poly(y2 + g * x + h,x)
```

```python
p_ = p
terms = []

s = [symbols(f's{i}') for i in range(5)]

for i in range(4,-1,-1):
    L   = LT(p_)
    p_ -= L
    terms.append(L/x**i - s[i])

terms
```

```python
solve(terms, a, b, d, e, f, g, h, dict=True)
```

```python
import numpy as np

u = np.random.rand(4,1)
v = np.random.rand(4,1)

A = u @ v.T

v *= np.linalg.norm(u)
u /= np.linalg.norm(u)

np.allclose(A, u @ v.T), np.allclose( A @ u, u * (u.T @ v) )
```

```python
np.linalg.eig(A)
```

```python
E = A - u * (u.T @ v) @ u.T 
np.linalg.eig( E.T @ E )
```
